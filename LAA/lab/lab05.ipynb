```python
import numpy as np
from scipy.io import loadmat

# Load the .mat file
file_path = "scores.mat"
data = loadmat(file_path)

# Extract the Scores matrix
A = np.array(data['Scores'], dtype=float)

# Define b as the sum of each team's row (net wins approximation)
b = np.sum(A, axis=1)

# Gaussian Elimination - Forward Elimination
def gaussian_elimination(A, b):
    n = len(b)
    for i in range(n):
        max_row = max(range(i, n), key=lambda r: abs(A[r, i]))
        if i != max_row:
            A[[i, max_row]] = A[[max_row, i]]
            b[i], b[max_row] = b[max_row], b[i]
        for j in range(i + 1, n):
            factor = A[j, i] / A[i, i]
            A[j, i:] -= factor * A[i, i:]
            b[j] -= factor * b[i]

    return A, b

# Back Substitution
def back_substitution(A, b):
    n = len(b)
    x = np.zeros(n)

    for i in range(n - 1, -1, -1):
        x[i] = (b[i] - np.dot(A[i, i+1:], x[i+1:])) / A[i, i]

    return x

# Solve the system
A, b = gaussian_elimination(A, b)
rankings = back_substitution(A, b)

# Normalize rankings for better readability
rankings = rankings - min(rankings)  # Shift so minimum is 0
rankings = rankings / max(rankings) * 100  # Scale to 100

# List of teams
teams = ["Oklahoma", "Oklahoma State", "West Virginia", "Kansas State", "TCU",
         "Baylor", "Texas", "Texas Tech", "Iowa State", "Kansas"]

# Sort teams by ranking (descending order)
ranked_teams = sorted(zip(teams, rankings), key=lambda x: x[1], reverse=True)

# Display final rankings
print("\nBig 12 Team Rankings (2016-2017):")
for rank, (team, score) in enumerate(ranked_teams, 1):
    print(f"{rank}. {team}: {score:.2f}")
```

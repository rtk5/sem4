```python
import numpy as np
import imageio.v3 as imageio
import matplotlib.pyplot as plt

image_path = "einstein.jpg"

# 1. Load an Image
def load_image(image_path):
    return imageio.imread(image_path)

# 2. Check dimensions
def get_image_size(image):
    return image.shape

# 3. Check if the array contains integer values
def is_integer_array(image):
    return np.issubdtype(image.dtype, np.integer)

# 4. Convert to Grayscale
def convert_to_grayscale(image):
    if image.ndim == 3:
        return np.mean(image, axis=2).astype(np.uint8)
    else:
        return image

# 5. Lightening & Darkening the Image
def adjust_brightness(image, factor):
    return np.clip(image * factor, 0, 255).astype(np.uint8)

# 6. Adjust Contrast
def adjust_contrast(image, factor):
    mean = np.mean(image)
    return np.clip((image - mean) * factor + mean, 0, 255).astype(np.uint8)

# 7. Crop the Image
def crop_image(image, x_start, x_end, y_start, y_end):
    return image[y_start:y_end, x_start:x_end]

# 8. Apply Matrix Transformations (Flip horizontally, Rotate 90 degrees)
def flip_image(image):
    return np.flip(image, axis=1)

def rotate_image(image):
    return np.rot90(image)

# 9. Display Image
def display_image(image, title="Image"):
    plt.imshow(image, cmap='gray')
    plt.title(title)
    plt.axis("off")
    plt.show()

# Example Usage
if __name__ == "__main__":
    img = load_image("einstein.jpg")
    print("Image Size:", get_image_size(img))
    print("Is Integer Array:", is_integer_array(img))

    gray_img = convert_to_grayscale(img)
    bright_img = adjust_brightness(gray_img, 1.5)
    contrast_img = adjust_contrast(gray_img, 1.8)
    cropped_img = crop_image(gray_img, 50, 200, 50, 200)
    flipped_img = flip_image(gray_img)
    rotated_img = rotate_image(gray_img)

    display_image(gray_img, "Grayscale Image")
    display_image(bright_img, "Brightened Image")
    display_image(contrast_img, "Contrast Adjusted Image")
    display_image(cropped_img, "Cropped Image")
    display_image(flipped_img, "Flipped Image")
    display_image(rotated_img, "Rotated Image")
```

```python
import numpy as np
from scipy.sparse import diags

# 1. Determine the length of a matrix or vector.
def matrix_length(matrix):
    return len(matrix)

# 2. Add B as the fourth row of A to create C.
def add_row(A, B):
    return np.vstack((A[:3], B, A[3:]))

# 3. Create D from rows 2, 3, 4 and columns 3, 4 of C.
def submatrix(C):
    return C[1:4, 2:4]

# 4. Transpose D to create E.
def transpose_matrix(D):
    return D.T

# 5. Check the size of E.
def matrix_size(matrix):
    return matrix.shape

# 6. Create equally spaced vectors.
def create_linspace(start, stop, num):
    return np.linspace(start, stop, num)

# 7 & 8. Find max/min in rows and columns of A.
def max_min_columns(A):
    return np.max(A, axis=0), np.min(A, axis=0)

def max_min_rows(A):
    return np.max(A, axis=1), np.min(A, axis=1)

# 9. Calculate mean and sum in each column and row of A.
def mean_sum(A):
    return np.mean(A, axis=0), np.sum(A, axis=0), np.mean(A, axis=1), np.sum(A, axis=1)

# 10. Create random matrices F and G.
F = np.random.randint(0, 10, (3, 3))
G = np.random.randint(0, 10, (3, 3))

# 11. Perform operations on F and G.
scalar = 2
scalar_mult = scalar * F
addition = F + G
subtraction = F - G
element_wise_mult = np.multiply(F, G)

# 12. Matrix multiplication if dimensions are compatible.
def matrix_multiply(A, F):
    if A.shape[1] == F.shape[0]:
        return np.dot(A, F)
    else:
        return "Incompatible dimensions"

# 13. Generate special matrices.
identity_matrix = np.eye(3)
zero_matrix = np.zeros((5, 3))
one_matrix = np.ones((4, 2))
diagonal_matrix = np.diag([1, 2, 7])
random_matrix = np.random.randint(0, 10, (6, 6))
diagonal_elements = np.diag(random_matrix)
sparse_diagonal_matrix = diags([1, 2, 7], [0, 1, 2], shape=(3, 3)).toarray()

def display_results():
    print("Identity Matrix:\n", identity_matrix)
    print("Zero Matrix:\n", zero_matrix)
    print("One Matrix:\n", one_matrix)
    print("Diagonal Matrix:\n", diagonal_matrix)
    print("Diagonal Elements from 6x6 Matrix:\n", diagonal_elements)
    print("Sparse Diagonal Matrix:\n", sparse_diagonal_matrix)

display_results()
```

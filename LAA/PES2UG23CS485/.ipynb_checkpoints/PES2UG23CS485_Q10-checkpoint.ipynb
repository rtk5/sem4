```python
#Making triangle

import numpy as np

t = np.linspace(0, 2 * np.pi, 4)
t = np.delete(t, 3)
v = np.array([np.cos(t), np.sin(t)])

#Making linear transformation matrix

T = np.array([[0.5, 0], [0, 0.5]])
x = np.random.rand(2, 1) - 0.5
```

```python
#Plotting triangles

import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 2 * np.pi, 4)[:-1]
v = np.array([np.cos(t), np.sin(t)])
T = np.array([[0.5, 0], [0, 0.5]])
x = np.random.rand(2, 1) - 0.5
Num = 10000
points = np.zeros((2, Num))
current_point = x
for i in range(Num):
    points[:, i] = current_point.flatten()
    vertex = v[:, np.random.choice(3)]
    current_point = T @ (current_point + vertex.reshape(2, 1))
plt.plot(v[0, :], v[1, :], 'k*', label='Vertices')
plt.plot(x[0, 0], x[1, 0], 'b.', label='Starting Point')
plt.plot(points[0, :], points[1, :], 'r.', markersize=1, label='Points')
plt.axis('equal')
plt.legend()
plt.show()
t = np.linspace(0, 2 * np.pi, 4)[:-1]
v = np.array([np.cos(t), np.sin(t)])
T = np.array([[0.5, 0], [0, 0.5]])
x = np.random.rand(2, 1) - 0.5
Num = 10000
points = np.zeros((2, Num + 1))
points[:, 0] = x.flatten()
for j in range(Num):
    k = np.random.randint(0, 3)
    current_point = points[:, j]
    transformed_point = T @ (current_point - v[:, k]) + v[:, k]
    points[:, j + 1] = transformed_point
plt.plot(points[0, :], points[1, :], 'b.', markersize=1)
plt.axis('equal')
plt.title('Sierpiński Triangle')
plt.show()
```

```python
#Square vertices fractal

t = np.linspace(0, 2 * np.pi, 5)[:-1]
v = np.array([np.cos(t), np.sin(t)])
T = np.array([[0.5, 0], [0, 0.5]])
x = np.random.rand(2, 1) - 0.5
Num = 10000
points = np.zeros((2, Num + 1))
points[:, 0] = x.flatten()
for j in range(Num):
    k = np.random.randint(0, 4)
    current_point = points[:, j]
    transformed_point = T @ (current_point - v[:, k]) + v[:, k]
    points[:, j + 1] = transformed_point
plt.figure()
plt.plot(points[0, :], points[1, :], 'b.', markersize=1)
plt.axis('equal')
plt.title('Fractal with Square Vertices')
plt.show()
t = np.linspace(0, 2 * np.pi, 5)[:-1]
v = np.array([np.cos(t), np.sin(t)])
T = np.array([[0.5, 0], [0, 0.5]])
x = np.random.rand(2, 1) - 0.5
Num = 10000
points = np.zeros((2, Num + 1))
points[:, 0] = x.flatten()
for j in range(Num):
    k = np.random.randint(0, 4)
    current_point = points[:, j]
    transformed_point = T @ (current_point - v[:, k]) + v[:, k]
    points[:, j + 1] = transformed_point
plt.figure()
plt.plot(points[0, :], points[1, :], 'b.', markersize=1)
plt.axis('equal')
plt.title('Fractal with Square Vertices')
plt.show()
```

```python
#Fractal with vertex preference

t = np.linspace(0, 2 * np.pi, 5)[:-1]
v = np.exp(1j * t)
T = np.array([[0.5, 0], [0, 0.5]])
x = np.random.rand(2, 1) - 0.5
Num = 5000
points = np.zeros((2, Num + 1))
points[:, 0] = x.flatten()
k1 = -1
for j in range(Num):
    k = np.random.randint(0, 4)
    if k >= k1:
        k = (k + 1) % 4
    w = np.array([np.real(v[k]), np.imag(v[k])])
    current_point = points[:, j]
    transformed_point = T @ (current_point - w) + w
    points[:, j + 1] = transformed_point
    k1 = k
plt.figure()
plt.plot(np.real(v), np.imag(v), 'k*', label='Vertices')
plt.plot(points[0, :], points[1, :], 'b.', markersize=1, label='Fractal Points')
plt.axis('equal')
plt.title('Fractal with Vertex Preference')
plt.legend()
plt.show()
```

```python
#Fractal with vertex preference which is no opposite vertex

t = np.linspace(0, 2 * np.pi, 5)[:-1]
v = np.exp(1j * t)
T = np.array([[0.5, 0], [0, 0.5]])
x = np.random.rand(2, 1) - 0.5
Num = 5000
points = np.zeros((2, Num + 1))
points[:, 0] = x.flatten()
k1 = 0
w = np.array([np.real(v[k1]), np.imag(v[k1])])
for j in range(Num):
    k = np.random.randint(0, 4)
    if (k != (k1 + 2) % 4) and ((k1 != (k + 2) % 4)):
        w = np.array([np.real(v[k]), np.imag(v[k])])
        transformed_point = T @ (points[:, j] - w) + w
        points[:, j + 1] = transformed_point
        k1 = k
    else:
        points[:, j + 1] = points[:, j]
plt.figure()
plt.plot(np.real(v), np.imag(v), 'k*', label='Vertices')
plt.plot(points[0, :], points[1, :], 'b.', markersize=1, label='Fractal Points')
plt.axis('equal')
plt.title('Fractal with Vertex Preference (No Opposite Vertex)')
plt.legend()
plt.show()
```

```python
#Barsnely fern

T1 = np.array([[0.85, 0.04], [-0.04, 0.85]])
T2 = np.array([[-0.15, 0.28], [0.26, 0.24]])
T3 = np.array([[0.2, -0.26], [0.23, 0.22]])
T4 = np.array([[0, 0], [0, 0.16]])
Q1 = np.array([0, 1.64])
Q2 = np.array([-0.028, 1.05])
Q3 = np.array([0, 1.6])
Q4 = np.array([0, 0])
P1 = 0.85
P2 = 0.07
P3 = 0.07
P4 = 0.01
Num = 15000
x = np.zeros((2, Num + 1))
x[:, 0] = np.random.rand(2)
plt.figure()
plt.plot(x[0, 0], x[1, 0], 'b.')
for j in range(Num):
    r = np.random.rand()
    if r <= P1:
        x[:, j + 1] = T1 @ x[:, j] + Q1
    elif r <= P1 + P2:
        x[:, j + 1] = T2 @ x[:, j] + Q2
    elif r <= P1 + P2 + P3:
        x[:, j + 1] = T3 @ x[:, j] + Q3
    else:
        x[:, j + 1] = T4 @ x[:, j] + Q4
plt.plot(x[0, :], x[1, :], 'b.', markersize=1)
plt.axis('equal')
plt.title('Barnsley Fern')
plt.show()
```

```python
#Fractal based on hexagon

t = np.linspace(0, 2 * np.pi, 7)[:-1]
v = np.array([np.cos(t), np.sin(t)])
T = np.array([[0.25, 0], [0, 0.25]])
Num = 15000
x = np.zeros((2, Num + 1))
x[:, 0] = np.random.rand(2) - 0.5
points = np.zeros((2, Num + 1))
points[:, 0] = x[:, 0]
for j in range(Num):
    k = np.random.randint(0, 6)
    points[:, j + 1] = T @ (points[:, j] - v[:, k]) + v[:, k]
plt.figure()
plt.plot(v[0, :], v[1, :], 'k*', label='Vertices')
plt.plot(points[0, :], points[1, :], 'b.', markersize=1, label='Fractal Points')
plt.axis('equal')
plt.title('Fractal Based on Regular Hexagon')
plt.legend()
plt.show()
```

```python
#Fractal based on pentagon with vertex restirction

t = np.linspace(0, 2 * np.pi, 6)[:-1]
v = np.array([np.cos(t), np.sin(t)])
T = np.array([[2/5, 0], [0, 2/5]])
Num = 10000
x = np.zeros((2, Num + 1))
x[:, 0] = np.random.rand(2) - 0.5
points = np.zeros((2, Num + 1))
points[:, 0] = x[:, 0]
prev_vertex_index = np.random.randint(0, 5)
for j in range(Num):
    current_vertex_index = prev_vertex_index
    while current_vertex_index == prev_vertex_index:
        current_vertex_index = np.random.randint(0, 5)
    w = v[:, current_vertex_index]
    points[:, j + 1] = T @ (points[:, j] - w) + w
    prev_vertex_index = current_vertex_index
plt.figure()
plt.plot(v[0, :], v[1, :], 'k*', label='Vertices')
plt.plot(points[0, :], points[1, :], 'b.', markersize=1, label='Fractal Points')
plt.axis('equal')
plt.title('Fractal Based on Regular Pentagon with Vertex Restriction')
plt.legend()
plt.show()
```
